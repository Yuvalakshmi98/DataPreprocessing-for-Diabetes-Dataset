```python
# Task 3 – Outlier Handling
# ● Retain Outliers: Retain outliers in the AGE, HbA1c, and BMI columns, explaining why
# they should be preserved.
import matplotlib.pyplot as plt
import seaborn as sns
cols = ['AGE', 'HbA1c', 'BMI']
plt.figure(figsize=(10, 5))
sns.boxplot(data=df[cols])
plt.title('Boxplot of AGE, HbA1c, and BMI')
plt.show()
```

```python
# Filter Outliers: In the Creatinine Ratio (Cr) column,
# exclude values that exceed the 99.5th percentile threshold. Similarly, in the Urea column, exclude values that
# exceed the 99.9th percentile threshold. Use these specific percentile
# thresholds for filtering, and do not apply IQR or other methods.

cret_threshold = df['Cr'].quantile(0.995)     # 99.5th percentile
urea_threshold = df['Urea'].quantile(0.999) # 99.9th percentile
print("Creatinine Threshold (99.5th percentile):", cret_threshold)
print("Urea Threshold (99.9th percentile):", urea_threshold)
```

```python
cr_exceed_threshold = df[df['Cr'] > cret_threshold]
urea_exceed_threshold = df[df['Urea'] > urea_threshold]
print("Rows with Cr exceeding threshold:", len(cr_exceed_threshold))
print("Rows with Urea exceeding threshold:", len(urea_exceed_threshold))
```

```python
df_filtered = df[(df['Cr'] < cret_threshold) &(df['Urea'] < urea_threshold)]
print("Original Shape:", df.shape)
print("Filtered Shape:", df_filtered.shape)
print("Rows Removed:", df.shape[0] - df_filtered.shape[0])
```

```python
# Remove Extreme Outliers: For lipid-related columns (LDL, VLDL, HDL, TG, Chol), remove
# extreme outliers using an appropriate method such as IQR or Z-score.
# Lipid-related columns
original_rows = len(df)
lipid_cols = ['LDL', 'VLDL', 'HDL', 'TG', 'Chol']
for col in lipid_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    df = df[(df[col] >= lower) & (df[col] <= upper)]
print("Before Handling outliers",original_rows)
print("After removing outliers",len(df))
print("No. of Removed rows:", original_rows - len(df))
df.reset_index(drop=True, inplace=True)
```
